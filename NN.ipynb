```python
from tensorflow import keras
from sklearn.neural_network import MLPClassifier
from keras.datasets import mnist 
import matplotlib.pyplot as plt 
import numpy as np
import random
import joblib

# Load and normalize data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train = x_train / 255
x_test = x_test / 255

# Flatten 28x28 images into 784 features
x_train_flat = x_train.reshape((x_train.shape[0], -1))
x_test_flat = x_test.reshape((x_test.shape[0], -1))

# Build and train the model
# One hidden layer (100 neurons) with max iters = 20 and verbose (print loss after each iteration)
model = MLPClassifier(hidden_layer_sizes=(100,), max_iter=1000, verbose=True) 
model.fit(x_train_flat, y_train)


# Evaluate accuracy
accuracy = model.score(x_test_flat, y_test)
print(f"Test Accuracy: {accuracy * 100:.2f}%")

joblib.dump(model, 'D:/ANU/Year 2/Second Term/Machine Learning/Assignments/2/mlp_model.pkl')

# Predict a random image
index = random.randint(0, len(x_test_flat) - 1)
image = x_test[index]
true_label = y_test[index]
predicted_label = model.predict([x_test_flat[index]])

# Show result
plt.imshow(image, cmap='gray')
plt.title(f"Predicted: {predicted_label[0]}, True: {true_label}")
plt.axis('off')
plt.show()
```
